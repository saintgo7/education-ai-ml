# YOLO 객체 탐지 (Object Detection)

YOLOv8을 사용한 실시간 객체 탐지를 학습합니다.

## 학습 목표
1. YOLO 아키텍처 이해
2. YOLOv8 모델 사용법
3. 커스텀 데이터셋 학습
4. 실시간 추론

```python
from ultralytics import YOLO
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import torch

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

## 1. 사전 학습된 YOLOv8 모델 로드

```python
# YOLOv8 모델 로드 (nano 버전)
model = YOLO('yolov8n.pt')

# 모델 정보
print(f"모델: YOLOv8 Nano")
print(f"클래스 수: {len(model.names)}")
print(f"\n클래스 목록:")
for i, name in model.names.items():
    print(f"  {i}: {name}")
```

## 2. 이미지 객체 탐지

```python
# 샘플 이미지로 테스트
# COCO 샘플 이미지 URL
image_url = 'https://ultralytics.com/images/bus.jpg'

# 추론
results = model(image_url)

# 결과 표시
for r in results:
    print(f"탐지된 객체 수: {len(r.boxes)}")
    
    for box in r.boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        xyxy = box.xyxy[0].tolist()
        print(f"  - {model.names[cls]}: {conf:.2%} at {xyxy}")
```

```python
# 결과 시각화
for r in results:
    im_array = r.plot()  # BGR numpy array
    im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(12, 8))
    plt.imshow(im_rgb)
    plt.axis('off')
    plt.title('YOLOv8 Object Detection')
    plt.show()
```

## 3. 다양한 YOLO 모델 비교

```python
# 다양한 크기의 YOLO 모델
model_variants = {
    'yolov8n': 'Nano (가장 빠름)',
    'yolov8s': 'Small',
    'yolov8m': 'Medium',
    'yolov8l': 'Large',
    'yolov8x': 'Extra Large (가장 정확)'
}

print("YOLOv8 모델 변형:")
for name, desc in model_variants.items():
    print(f"  {name}.pt - {desc}")

print("\n태스크별 모델:")
print("  yolov8n.pt - 객체 탐지")
print("  yolov8n-seg.pt - 인스턴스 세그멘테이션")
print("  yolov8n-pose.pt - 포즈 추정")
print("  yolov8n-cls.pt - 이미지 분류")
```

## 4. 커스텀 데이터셋 학습

```python
# 커스텀 데이터셋 YAML 예시
yaml_content = """
# 커스텀 데이터셋 설정 예시
path: /path/to/dataset  # 데이터셋 루트 경로
train: images/train      # 학습 이미지 경로
val: images/val          # 검증 이미지 경로
test: images/test        # 테스트 이미지 경로 (선택)

# 클래스
names:
  0: person
  1: car
  2: bicycle
"""
print("커스텀 데이터셋 YAML 형식:")
print(yaml_content)
```

## 5. 모델 검증 및 내보내기

```python
# 모델 검증
# metrics = model.val()
# print(f"mAP50: {metrics.box.map50}")
# print(f"mAP50-95: {metrics.box.map}")

print("모델 검증 코드:")
print("""
metrics = model.val()
print(f"mAP50: {metrics.box.map50}")
print(f"mAP50-95: {metrics.box.map}")
""")

print("\n모델 내보내기 코드:")
print("""
# ONNX로 내보내기
model.export(format='onnx')

# TensorRT로 내보내기
model.export(format='engine')

# CoreML로 내보내기 (iOS)
model.export(format='coreml')
""")
```

## 6. 실시간 비디오 처리

```python
# 웹캠 실시간 객체 탐지 코드
video_code = """
import cv2
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
cap = cv2.VideoCapture(0)  # 웹캠

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # 추론
    results = model(frame)
    
    # 결과 시각화
    annotated_frame = results[0].plot()
    
    cv2.imshow('YOLOv8 Detection', annotated_frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
"""

print("실시간 비디오 처리 코드:")
print(video_code)
```
