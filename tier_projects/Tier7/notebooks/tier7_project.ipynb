## 3. 데이터 탐색

데이터의 특성과 분포를 파악합니다.

```python
# 샘플 데이터 생성
n_samples = 100
n_features = 5

X = np.random.randn(n_samples, n_features)
y = (X[:, 0] + X[:, 1] > 0).astype(int)

# 데이터 요약
print(f'데이터 형태: X={X.shape}, y={y.shape}')
print(f'클래스 분포: 0={np.sum(y==0)}, 1={np.sum(y==1)}')
print(f'특성 평균: {X.mean(axis=0)}')
print(f'특성 표준편차: {X.std(axis=0)}')
```

## 4. 모델 구현 및 학습

Tier에서 배운 알고리즘을 구현하고 학습시킵니다.

```python
# 모델 구현 (기본 분류기)
class SimpleClassifier:
    def __init__(self):
        self.weights = None
        self.bias = None
    
    def fit(self, X, y):
        # 간단한 학습 로직
        self.weights = np.mean(X[y==1], axis=0) - np.mean(X[y==0], axis=0)
        self.bias = 0
        return self
    
    def predict(self, X):
        scores = X @ self.weights + self.bias
        return (scores > 0).astype(int)

# 모델 학습
model = SimpleClassifier()
model.fit(X, y)
y_pred = model.predict(X)

accuracy = np.mean(y_pred == y)
print(f'모델 정확도: {accuracy:.4f}')
```

## 5. 모델 평가

다양한 지표로 모델 성능을 평가합니다.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# 성능 지표 계산
precision = precision_score(y, y_pred, zero_division=0)
recall = recall_score(y, y_pred, zero_division=0)
f1 = f1_score(y, y_pred, zero_division=0)

print(f'평가 결과:')
print(f'  정확도: {accuracy:.4f}')
print(f'  정밀도: {precision:.4f}')
print(f'  재현율: {recall:.4f}')
print(f'  F1 스코어: {f1:.4f}')
```

## 6. 결과 시각화

모델 예측 결과를 시각화합니다.

```python
# 혼동 행렬
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y, y_pred)

fig, ax = plt.subplots(1, 1, figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax)
ax.set_title(f'Tier {tier_num} 프로젝트 - 혼동 행렬')
ax.set_ylabel('실제')
ax.set_xlabel('예측')
plt.tight_layout()
plt.show()
```
